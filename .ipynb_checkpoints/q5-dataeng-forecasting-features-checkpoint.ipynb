```python
import pandas as pd
import numpy as np

class SalesData:
    def __init__(self, sales_path, product_path, brand_path, store_path):
        # Load the data
        self.sales_df = pd.read_csv(sales_path)
        self.product_df = pd.read_csv(product_path)
        self.brand_df = pd.read_csv(brand_path)
        self.store_df = pd.read_csv(store_path)
        print("Data loaded.")
        print("Sales DataFrame:")
        print(self.sales_df.head())

    def preprocess_data(self, min_date, max_date):
        # Filter by date range
        self.sales_df['date'] = pd.to_datetime(self.sales_df['date'])
        print(f"\nDate column converted and filtered by date range: {min_date} - {max_date}")
        self.sales_df = self.sales_df[(self.sales_df['date'] >= min_date) & (self.sales_df['date'] <= max_date)]
        print("Filtered Sales DataFrame:")
        print(self.sales_df.head())

        # Merge product and brand info
        self.sales_df = pd.merge(self.sales_df, self.product_df[['id', 'name', 'brand']], left_on='product', right_on='id', how='left')
        self.sales_df.rename(columns={'id': 'product_id', 'brand': 'brand_id'}, inplace=True)
        print("\nSales DataFrame merged with product and brand info:")
        print(self.sales_df.head())

class SalesAnalysis:
    def __init__(self, sales_data):
        self.sales_data = sales_data

    def calculate_moving_averages_and_lag(self):
        print("\nCalculating store-level sales...")
        # Calculate store-level sales
        self.store_sales = self.sales_data.sales_df.groupby(['store', 'date'])['quantity'].sum().reset_index()
        self.store_sales['MA7_S'] = self.store_sales.groupby(['store'])['quantity'].transform(lambda x: x.rolling(7, min_periods=1).mean())
        self.store_sales['LAG7_S'] = self.store_sales.groupby(['store'])['quantity'].shift(7)
        print("Store-level sales:")
        print(self.store_sales.head())

        print("\nCalculating brand-level sales...")
        # Calculate brand-level sales
        self.brand_sales = self.sales_data.sales_df.groupby(['store', 'brand_id', 'date'])['quantity'].sum().reset_index()
        self.brand_sales['MA7_B'] = self.brand_sales.groupby(['store', 'brand_id'])['quantity'].transform(lambda x: x.rolling(7, min_periods=1).mean())
        self.brand_sales['LAG7_B'] = self.brand_sales.groupby(['store', 'brand_id'])['quantity'].shift(7)
        print("Brand-level sales:")
        print(self.brand_sales.head())

        print("\nCalculating product-level sales...")
        # Calculate product-level sales
        self.sales_data.sales_df['MA7_P'] = self.sales_data.sales_df.groupby(['store', 'product_id'])['quantity'].transform(lambda x: x.rolling(7, min_periods=1).mean())
        self.sales_data.sales_df['LAG7_P'] = self.sales_data.sales_df.groupby(['store', 'product_id'])['quantity'].shift(7)
        print("Product-level sales:")
        print(self.sales_data.sales_df[['product_id', 'store', 'quantity', 'MA7_P', 'LAG7_P']].head())

class WMAPE:
    @staticmethod
    def calculate(forecast, actual):
        if actual == 0:  # Skip WMAPE calculation if actual is zero
            return 0
        return abs(forecast - actual) / actual * 100

class FeatureEngineering:
    def __init__(self, sales_data, store_sales, brand_sales):
        self.sales_data = sales_data
        self.store_sales = store_sales
        self.brand_sales = brand_sales

    def generate_features(self):
        print("\nGenerating features...")
        # Merge data
        final_df = pd.merge(self.sales_data.sales_df, self.brand_sales[['store', 'brand_id', 'date', 'MA7_B', 'LAG7_B']], on=['store', 'brand_id', 'date'], how='left')
        final_df = pd.merge(final_df, self.store_sales[['store', 'date', 'MA7_S', 'LAG7_S']], on=['store', 'date'], how='left')
        
        print("Merged data:")
        print(final_df.head())
        
        # Create features file (features.csv)
        features_df = final_df[['product_id', 'store', 'brand_id', 'date', 'MA7_P', 'LAG7_P', 'MA7_B', 'LAG7_B', 'MA7_S', 'LAG7_S']]
        features_df.to_csv('features.csv', index=False)
        print("Features saved to 'features.csv'.")
        return final_df

class WMAPEAnalysis:
    def __init__(self, final_df):
        self.final_df = final_df

    def calculate_wmape(self):
        print("\nCalculating WMAPE...")
        wmape_df = self.final_df[['product_id', 'store', 'brand_id', 'MA7_P', 'quantity']].copy()
        wmape_df['WMAPE'] = wmape_df.apply(lambda row: WMAPE.calculate(row['MA7_P'], row['quantity']), axis=1)
        print("Calculated WMAPE data:")
        print(wmape_df.head())
        return wmape_df

    def get_top_wmape(self, wmape_df, top):
        print("\nSorting WMAPE results...")
        wmape_df = wmape_df.groupby(['product_id', 'store', 'brand_id'], as_index=False).agg({'WMAPE': 'mean'})
        wmape_df = wmape_df.sort_values(by='WMAPE', ascending=False).head(top)
        wmape_df.to_csv('mapes.csv', index=False)
        print("Top WMAPE results saved to 'mapes.csv'.")
        return wmape_df

def main(min_date, max_date, top, sales_path, product_path, brand_path, store_path):
    print("\nLoading and processing data...")
    # Load and process the data
    sales_data = SalesData(sales_path, product_path, brand_path, store_path)
    sales_data.preprocess_data(min_date, max_date)
    
    # Perform sales analysis and calculations
    sales_analysis = SalesAnalysis(sales_data)
    sales_analysis.calculate_moving_averages_and_lag()

    # Perform feature engineering
    feature_engineering = FeatureEngineering(sales_data, sales_analysis.store_sales, sales_analysis.brand_sales)
    final_df = feature_engineering.generate_features()

    # Perform WMAPE calculation and sorting
    wmape_analysis = WMAPEAnalysis(final_df)
    wmape_df = wmape_analysis.calculate_wmape()
    top_wmape = wmape_analysis.get_top_wmape(wmape_df, top)

    return top_wmape

# Run the code
min_date = "2021-01-08"
max_date = "2021-05-30"
top = 5
sales_path = '/Users/silakazan/Desktop/data/sales.csv'
product_path = '/Users/silakazan/Desktop/data/product.csv'
brand_path = '/Users/silakazan/Desktop/data/brand.csv'
store_path = '/Users/silakazan/Desktop/data/store.csv'

wmape_results = main(min_date, max_date, top, sales_path, product_path, brand_path, store_path)

# Print the WMAPE results
print("\nFinal WMAPE Results:")
print(wmape_results)
```
