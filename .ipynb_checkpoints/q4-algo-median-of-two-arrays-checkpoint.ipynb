```python
#Question_2

def merge_sorted_arrays(arr1, arr2):
    # Merge two sorted arrays without using built-in sort
    merged = []
    i, j = 0, 0

    # Merge elements until one of the arrays is fully traversed
    while i < len(arr1) and j < len(arr2):
        if arr1[i] <= arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1

    # Append remaining elements from arr1 or arr2
    while i < len(arr1):
        merged.append(arr1[i])
        i += 1

    while j < len(arr2):
        merged.append(arr2[j])
        j += 1

    return merged


def calculate_median(arr):
    # Calculate the median of a sorted array
    n = len(arr)
    if n % 2 == 1:
        # Odd length: return the middle element
        return arr[n // 2]
    else:
        # Even length: return the average of two middle elements
        mid1, mid2 = arr[(n // 2) - 1], arr[n // 2]
        return (mid1 + mid2) / 2


if __name__ == "__main__":
    # Input from the command line
    input1 = list(map(float, input("input1: ").strip().split()))
    input2 = list(map(float, input("input2: ").strip().split()))

    # Merge the arrays and calculate median
    merged_array = merge_sorted_arrays(input1, input2)
    median = calculate_median(merged_array)

    # Output the result
    print(f"median: {median}")
```
