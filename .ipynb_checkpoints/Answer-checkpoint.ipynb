```python
import pandas as pd

# ------------------
# Question 1a: Data Loading and Initial Inspection
# ------------------

# File paths

file_paths = {
    "BAXTER": "/Users/silakazan/Desktop/Bilkent/BAXTER copy 2.xlsx",
    "BEECH": "/Users/silakazan/Desktop/Bilkent/BEECH.xlsx",
    "BITTERN": "/Users/silakazan/Desktop/Bilkent/BITTERN.xlsx",
    "CAVENDISH": "/Users/silakazan/Desktop/Bilkent/CAVENDISH.xlsx",
    "OAKDAT_subject2": "/Users/silakazan/Desktop/Bilkent/OAKDAT_subject2.xlsx",
    "FOWLER": "/Users/silakazan/Desktop/Bilkent/FOWLER.xlsx",
    "LAMMAS": "/Users/silakazan/Desktop/Bilkent/LAMMAS.xlsx",
    "LIME_TREE": "/Users/silakazan/Desktop/Bilkent/LIME_TREE.xlsx",
    "LOPAR_subject2": "/Users/silakazan/Desktop/Bilkent/LOPAR_subject2.xlsx",
    "CORN_PLACE": "/Users/silakazan/Desktop/Bilkent/CORN_PLACE.xlsx"

}

# Load all files
print("Loading all files...")
dataframes = {}
for name, path in file_paths.items():
    try:
        df = pd.read_excel(path, dtype=str)  # Load all variables as strings
        dataframes[name] = df
        print(f"{name} successfully loaded! Shape: {df.shape}")
    except Exception as e:
        print(f"Error: {name} file could not be loaded - {e}\n")

# Display the first 5 rows of each file
for name, df in dataframes.items():
    print(f"\nFirst 5 rows - {name}:")
    print(df.head().to_string(index=False))

```

```python
# ------------------
# Question 1b: Identifying Common and Unique Variables
# ------------------

# Compare OAKDAT and LOPAR datasets
oakdat_columns = set(dataframes["BAXTER"].columns)
lopar_columns = set(dataframes["BEECH"].columns)

common_columns = oakdat_columns.intersection(lopar_columns)
oakdat_unique_columns = oakdat_columns - common_columns
lopar_unique_columns = lopar_columns - common_columns

print("\nCommon variables:")
print(common_columns)

print("\nVariables unique to OAKDAT:")
print(oakdat_unique_columns)

print("\nVariables unique to LOPAR:")
print(lopar_unique_columns)
```

```python
# ------------------
# Question 1c: Merging Student Files
# ------------------

oakdat_student_files = ["BAXTER", "BITTERN", "LAMMAS"]
lopar_student_files = ["BEECH", "CAVENDISH", "CORN_PLACE", "FOWLER", "LIME_TREE"]

for file in oakdat_student_files:
    df = dataframes[file]
    df.rename(columns={
        "subject_info": "subject",
        "sss_site": "ssssite"
    }, inplace=True)
    dataframes[file] = df

# Merge OAKDAT and LOPAR student datasets
student_data = pd.concat([dataframes[file] for file in oakdat_student_files + lopar_student_files], ignore_index=True)
print(f"Student data merged! New shape: {student_data.shape}")
```

```python

# ------------------
# Question 1d: Merging Teacher Data
# ------------------

teacher_lopar = dataframes["LOPAR_subject2"]
teacher_oakdat = dataframes["OAKDAT_subject2"]
teacher_oakdat.rename(columns={"sss_site": "ssssite"}, inplace=True)

teacher_data = pd.concat([teacher_lopar, teacher_oakdat], ignore_index=True)
full_data = student_data.merge(teacher_data, on=["site", "ssite", "ssssite"], how="left")
print(f"Teacher data added! New shape: {full_data.shape}")
```

```python
# ------------------
# Question 1e: Merging Time 2 Data
# ------------------ 

lopar_time2 = pd.read_excel("/Users/silakazan/Desktop/Bilkent/LOPAR_subject2.xlsx", dtype=str)
lopar_time2.rename(columns={"subject_info": "subject", "sss_site": "ssssite"}, inplace=True)
full_data_time2 = full_data.merge(lopar_time2, on=["site", "ssite", "ssssite"], how="left")
print(f"Time 2 data added! New shape: {full_data_time2.shape}")
```

```python
# ------------------
# Question 1f & 1g: Checking for Missing and Incorrect Data & Creating f_status
# ------------------

# Compare Time 1 and Time 2 student lists
students_time1 = set(full_data["subject"].dropna())
students_time2 = set(lopar_time2["subject2"].dropna())

# Determine student movements
status_mapping = {}
for student in students_time1:
    if student in students_time2:
        status_mapping[student] = "Continuing"
    else:
        status_mapping[student] = "Left"

for student in students_time2:
    if student not in students_time1:
        status_mapping[student] = "New"

# Add `f_status` variable to the student dataset
full_data["f_status"] = full_data["subject"].map(status_mapping)
print("\nStudent movements added with the f_status variable!")

print("\nAll steps successfully completed!")
```

```python
# ------------------
# Question 2a: Calculating Student Ages (Fixed Errors)
# ------------------

# Replace invalid values like "NOT IN LIST" with NaN
full_data["year"] = pd.to_numeric(full_data["year"], errors="coerce")

# Fill missing values with the most common year (mode)
most_common_year = full_data["year"].mode()[0]
full_data.loc[:, "year"] = full_data["year"].fillna(most_common_year)

# Calculate student ages
full_data["student_age"] = 2024 - full_data["year"]
print("\nStudent ages successfully calculated!")
print(full_data[["subject", "year", "student_age"]].head())
```

```python
# ------------------
# Question 2b: Transforming the Gender Variable
# ------------------

full_data["gender"] = full_data["dem1"].replace({"MALE": "Male", "FEMALE": "Female", "ERKEK": "Male", "KIZ": "Female"})
print("\nGender variable adjusted!")
print(full_data[["subject", "gender"]].head())
```

```python
# ------------------
# Question 2c: Adjusting Teacher Ages and Marital Status
# ------------------

teacher_age_columns = ["subject2_var1"]  # Variable assumed to represent teacher age
full_data["teacher_age"] = full_data[teacher_age_columns[0]].astype(float)
full_data["teacher_marital_status"] = full_data["subject2_var10"].replace({"EVLI": "Married", "BEKAR": "Single"})
print("\nTeacher age and marital status updated!")
print(full_data[["subject", "teacher_age", "teacher_marital_status"]].head())
```

```python
# ------------------
# Question 2d: Standardizing Student Test Scores
# ------------------

test_columns = [col for col in full_data.columns if "_test_" in col]
full_data[test_columns] = full_data[test_columns].apply(pd.to_numeric, errors='coerce')
full_data[test_columns] = (full_data[test_columns] - full_data[test_columns].mean()) / full_data[test_columns].std()
print("\nAll student test scores have been standardized!")
print(full_data[["subject"] + test_columns[:5]].head())
```

```python
# ------------------
# Question 2e: Checking and Filling Missing Data
# ------------------

missing_values = full_data.isnull().sum()
print("\nNumber of missing values:")
print(missing_values[missing_values > 0])

# Filling missing values (using mean or mode)
for col in full_data.columns:
    if full_data[col].dtype == "float64":
        full_data[col].fillna(full_data[col].mean(), inplace=True)
    elif full_data[col].dtype == "object":
        full_data[col].fillna(full_data[col].mode()[0], inplace=True)
print("\nMissing values have been filled!")
print(full_data.head())
```

```python
# ------------------
# Question 2f: Transforming Teacher and Student Survey Variables
# ------------------

survey_cols = [col for col in full_data.columns if col.startswith("a_s")]  # Student survey questions
full_data[survey_cols] = full_data[survey_cols].apply(pd.to_numeric, errors='coerce')
print("\nAll survey variables have been converted to numeric!")
print(full_data[["subject"] + survey_cols[:5]].head())
```

```python
# ------------------
# Question 2g: Identifying Missing Categorical Variables
# ------------------

categorical_columns = ["gender", "teacher_marital_status", "f_status"]
for col in categorical_columns:
    print(f"Categories of {col} variable:")
    print(full_data[col].value_counts())

print("\nAll Question 2 steps have been completed!")
```

```python
# Question 2h: Converting Survey Variables to Numeric (Both Student and Teacher)
# ------------------

survey_cols = [col for col in full_data.columns if col.startswith("a_s") or col.startswith("subject2_var")]  # Student and teacher survey questions
full_data[survey_cols] = full_data[survey_cols].apply(pd.to_numeric, errors='coerce')
print("\nSurvey variables successfully converted to numeric!")
print(full_data[survey_cols].head())
```

```python
# ------------------
# Question 2i: Creating Dummy Variables for f_choice
# ------------------

if "f_choice" in full_data.columns:
    dummy_vars = pd.get_dummies(full_data["f_choice"], prefix="choice", drop_first=True)
    selected_dummies = dummy_vars.iloc[:, :8]  # Selecting the first 8 dummy variables
    full_data = pd.concat([full_data, selected_dummies], axis=1)
    print("\n8 dummy variables created!")
    print(selected_dummies.head())
else:
    print("\nError: 'f_choice' variable not found. Needs to be checked!")

print("\nAll Question 2 steps have been completed!")
```

```python

import matplotlib.pyplot as plt
import seaborn as sns
# ------------------
# Question 3a: Basic Descriptive Statistics
# ------------------

descriptive_stats = full_data.describe()
print("\nBasic Descriptive Statistics:")
print(descriptive_stats)

```

```python
# ------------------
# Question 3b: Visualizing Gender Distribution
# ------------------

plt.figure(figsize=(6,4))
sns.countplot(x=full_data['gender'], palette='pastel')
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")
plt.show()
```

```python

# ------------------
# Question 3c: Student Age Distribution
# ------------------

plt.figure(figsize=(8,5))
sns.histplot(full_data['student_age'], bins=10, kde=True)
plt.title("Student Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Students")
plt.show()
```

```python
# ------------------
# Question 3d: Examining Correlations Between Test Scores
# ------------------

test_columns = [col for col in full_data.columns if "_test_" in col]
correlation_matrix = full_data[test_columns].corr()
plt.figure(figsize=(12, 10))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title("Test Scores Correlation Matrix")
plt.show()
```

```python
# ------------------
# Question 3e: Relationship Between Teacher Age and Student Performance
# ------------------

plt.figure(figsize=(8,6))
sns.scatterplot(x=full_data['teacher_age'], y=full_data['a_test_01'], alpha=0.6)
sns.regplot(x=full_data['teacher_age'], y=full_data['a_test_01'], scatter=False, color='red')
plt.title("Relationship Between Teacher Age and Test Score")
plt.xlabel("Teacher Age")
plt.ylabel("Student Test Score")
plt.show()
```

```python
# ------------------
# Question 3f: Distribution of Test Scores by Gender
# ------------------

plt.figure(figsize=(8,6))
sns.boxplot(x=full_data['gender'], y=full_data['a_test_01'], palette='coolwarm')
plt.title("Distribution of Test Scores by Gender")
plt.xlabel("Gender")
plt.ylabel("Test Score")
plt.show()

print("\nQuestion 3 steps successfully completed!")
```
